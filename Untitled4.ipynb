```python
df["Salary_Clean"] = df["Salary"]
```

```python
df.loc[df["Salary_Clean"].str.contains("month", na=False), "Salary_Clean"].head(10)
```

```python
df["Salary_Clean"] = df["Salary_Clean"].str.replace("₹", "", regex=False)
```

```python
df["Salary_Clean"] = df["Salary_Clean"].str.replace(",", "", regex=False)
```

```python
df["Salary_Clean"] = df["Salary_Clean"].str.replace("a month", "", regex=False)
```

```python
df["Salary_Clean"] = df["Salary_Clean"].str.replace("a day", "", regex=False)
```

```python
df["Salary_Clean"] = df["Salary_Clean"].str.replace("an hour", "", regex=False)
```

```python
df["Salary_Clean"].head(10)
```

```python
df["Salary_Clean"] = df["Salary_Clean"].str.replace(" a year", "", regex=False)
```

```python
df["Salary_Clean"].head(10)
```

```python
df["Salary_Min"] = df["Salary_Clean"].str.extract(r'(\d+)')[0]
```

```python
df["Salary_Max"] = df["Salary_Clean"].str.extract(r'-\s*(\d+)')[0]
```

```python
df[["Salary_Clean", "Salary_Min", "Salary_Max"]].head(10)
```

```python
df["Salary_Min"] = pd.to_numeric(df["Salary_Min"])
```

```python
df["Salary_Max"] = pd.to_numeric(df["Salary_Max"])
```

```python
df.loc[df["Salary"].str.contains("month", na=False), "Salary_Min_Annual"] = (
    df.loc[df["Salary"].str.contains("month", na=False), "Salary_Min"] * 12
)
```

```python
df.loc[df["Salary"].str.contains("month", na=False), "Salary_Max_Annual"] = (
    df.loc[df["Salary"].str.contains("month", na=False), "Salary_Max"] * 12
)
```

```python
df.loc[df["Salary"].str.contains("year", na=False), "Salary_Min_Annual"] = (
    df.loc[df["Salary"].str.contains("year", na=False), "Salary_Min"]
)
```

```python
df.loc[df["Salary"].str.contains("year", na=False), "Salary_Max_Annual"] = (
    df.loc[df["Salary"].str.contains("year", na=False), "Salary_Max"]
)
```

```python
df[["Salary", "Salary_Min_Annual", "Salary_Max_Annual"]].head(15)
```

```python
df.loc[df["Salary"].str.contains("day", na=False), "Salary_Min_Annual"] = (
    df.loc[df["Salary"].str.contains("day", na=False), "Salary_Min"] * 365
)
```

```python
df.loc[df["Salary"].str.contains("day", na=False), "Salary_Max_Annual"] = (
    df.loc[df["Salary"].str.contains("day", na=False), "Salary_Max"] * 365
)
```

```python
df.loc[df["Salary"].str.contains("hour", na=False), "Salary_Min_Annual"] = (
    df.loc[df["Salary"].str.contains("hour", na=False), "Salary_Min"] * 8 * 365
)
```

```python
df.loc[df["Salary"].str.contains("hour", na=False), "Salary_Max_Annual"] = (
    df.loc[df["Salary"].str.contains("hour", na=False), "Salary_Max"] * 8 * 365
)
```

```python
df["Annual_Salary"] = df[["Salary_Min_Annual", "Salary_Max_Annual"]].mean(axis=1)
```

```python
df["Salary_Category"] = pd.cut(
    df["Annual_Salary"],
    bins=[0, 200000, 300000, 500000, float("inf")],
    labels=["Below 2L", "2L-3L", "3L-5L", "Above 5L"]
)

df["Salary_Category"].value_counts()
```

```python
import matplotlib.pyplot as plt

df["Salary_Category"].value_counts().plot(kind="bar")

plt.title("Job Distribution by Salary Category")
plt.xlabel("Salary Category")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=0)
plt.show()
```

```python
state_counts = df[df["State"] != "Remote"]["State"].value_counts().head(10)

state_counts
```

```python
job_salary = (
    df.dropna(subset=["Annual_Salary"])
      .groupby("Job Title")
      .agg(
          Job_Count=("Job Title", "count"),
          Average_Salary=("Annual_Salary", "mean")
      )
)

job_salary = job_salary[job_salary["Job_Count"] >= 3]

job_salary.sort_values("Average_Salary", ascending=False).head(10)
```

```python
top_jobs = job_salary.sort_values(
    "Average_Salary",
    ascending=False
).head(10)

top_jobs["Average_Salary"].sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Job Titles by Average Salary")
plt.xlabel("Average Annual Salary (₹)")
plt.ylabel("Job Title")
plt.show()
```

```python
data_analyst_jobs = df[
    df["Job Title"].str.contains("Data Analyst", case=False, na=False)
]

data_analyst_jobs[
    ["Job Title", "Location", "State", "Annual_Salary"]
]
```

```python
data_analyst_jobs["Location"].value_counts().plot(kind="bar")

plt.title("Data Analyst Jobs by Location")
plt.xlabel("Location")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
salary_data_analyst = data_analyst_jobs.dropna(
    subset=["Annual_Salary"]
)

salary_data_analyst[
    ["Job Title", "Location", "Annual_Salary"]
]
```

```python
salary_data_analyst["Annual_Salary"].plot(
    kind="bar"
)

plt.title("Data Analyst Salary by Location")
plt.xlabel("Job Listing")
plt.ylabel("Annual Salary (₹)")
plt.show()
```

```python
salary_data_analyst = salary_data_analyst.copy()

salary_data_analyst.plot(
    x="Location",
    y="Annual_Salary",
    kind="bar",
    legend=False
)

plt.title("Data Analyst Salary by Location")
plt.xlabel("Location")
plt.ylabel("Annual Salary (₹)")
plt.xticks(rotation=45, ha="right")
plt.show()
```
